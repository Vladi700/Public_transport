```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def exponential_decay(t):
    return np.exp((-0.9) * t) / sum(np.exp((-0.9) * t))
```

```python
plt.figure(figsize=(8, 5))
t = np.linspace(0, 25, 100)
tau = 7

plt.plot(t, exponential_decay(t), label=f'tau={tau}')
plt.title('Exponential Decay Function')
plt.xlabel('Time (t)')
plt.ylabel('f(t)')
plt.legend()
plt.grid()
plt.show()
```

```python
sum(exponential_decay(t))
```

```python
def something(t):
    return (1 - exponential_decay(t))
```

```python
plt.figure(figsize=(8, 5))
plt.plot(t, something(t), label=f'tau={tau}')   
```
